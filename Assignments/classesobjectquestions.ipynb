### Assignment 10: Class with Properties

Create a class named `Rectangle` with private attributes `length` and `width`. Use properties to get and set these attributes. Create an object of the class and test the properties.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    @property
    def getlength(self):
        return self.__length

    @getlength.setter
    def setlength(self, length):
        self.__length = length

    @property
    def getwidth(self):
        return self.__width

    @getwidth.setter
    def setwidth(self, width):
        self.__width = width

rect = Rectangle(10, 5)
print(rect.getlength, rect.getwidth)  # 10 5
```

### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        print("This is area method from Shape class")
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius**2

class Square(Shape):
    def __init__(self,length):
        self.length = length

    def area(self):
        return self.length**2

circle = Circle(5)
square = Square(4)
print(circle.area())  # 78.53981633974483
print(square.area())  # 16
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self, x , y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y+other.y)

    def __str__(self):
        return f"vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2
print(v3)  # Vector(6, 8)
```

### Assignment 13: Class with Custom Exception

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python
class InsufficientBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self,account_number,balance = 0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self,ammout):
        self.balance += ammout
        

    def withdraw(self,amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient Balance")
        else:
            self.balance -= amount

    def getBalance(self):
        return self.balance


account1 = BankAccount('aer2000',4000)
account1.getBalance()
account1.deposit(10000)
try:
    account1.withdraw(20000)
except InsufficientBalanceError as er:
    print(f"Error: {er}")

account1.getBalance()

        
```

### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file
    def __exit__(self,excType, excValue, traceback):
        self.file.close()

with FileManager('fileHandlingAssignment/sample.txt', 'r') as file:
    content = file.read()
    print(content)
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
    def __init__(self,value = 0):
        self.value = value
    
    def add(self, amount):
        self.value += amount
        return self   ## self is retured to do the method chaining. 


    def subtract(self, amount):
        self.value -= amount
        return self

    def multiply(self, amount):
            self.value *= amount
            return self

    def divide(self, amount):
            if amount !=0:
                self.value /= amount
                return self
            else:
                 print("cannot be divided by 0")

calc = Calculator()
calc.add(10).subtract(3).multiply(20).divide(2)
calc.add(10)
# calc.divide(10)
# calc.divide(0)
print(calc.value)  # 70
```
